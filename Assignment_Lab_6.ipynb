```python
import gymnasium as gym
import numpy as np
```

```python
def random_agent(env_name, episodes=5):

    print("=" * 60)
    print("Environment:", env_name)
    print("=" * 60)

    # Create environment
    env = gym.make(env_name)

    episode_rewards = []

    # Run multiple episodes
    for episode in range(1, episodes + 1):

        # Reset environment
        state, info = env.reset()

        total_reward = 0
        steps = 0

        terminated = False
        truncated = False

        # Continue until episode terminates
        while not (terminated or truncated):

            # Select a random action
            action = env.action_space.sample()

            # Execute action
            next_state, reward, terminated, truncated, info = env.step(action)

            # Update reward and steps
            total_reward += reward
            steps += 1

            # Move to next state
            state = next_state

        # Calculate average reward
        average_reward = total_reward / steps

        episode_rewards.append(total_reward)

        print(f"Episode {episode}")
        print(f"Steps          : {steps}")
        print(f"Total Reward   : {total_reward}")
        print(f"Average Reward : {average_reward:.2f}")
        print("-" * 40)

    # Display overall result
    print("\nTotal reward for each episode:")
    print(episode_rewards)

    print("Average Total Reward:",
          np.mean(episode_rewards))

    env.close()
```

```python
random_agent("CliffWalking-v1", episodes=5)
```

```python
random_agent("Taxi-v4", episodes=5)
```
